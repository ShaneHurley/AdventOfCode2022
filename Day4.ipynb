```
input = open("day4.txt", "r")
```

```
pairs = input.read().split("\n")
pairs = list(filter(lambda x: len(x) > 0, pairs))
```

```
# part 1: one range fully contains the other
total = 0
for pair in pairs:
    left, right = pair.split(",")
    a1, a2 = map(int, left.split("-"))
    b1, b2 = map(int, right.split("-"))
    if (a1 <= b1 and a2 >= b2) or (b1 <= a1 and b2 >= a2):
        total += 1
print(total)
```

```
# part 2: ranges overlap at all (even a single section)
# Only the if-condition changes from part 1:
#   full contain -> a1 <= b2 and b1 <= a2
total = 0
for pair in pairs:
    left, right = pair.split(",")
    a1, a2 = map(int, left.split("-"))
    b1, b2 = map(int, right.split("-"))
    if a1 <= b2 and b1 <= a2:
        total += 1
print(total)
```
