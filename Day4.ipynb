```
input = open("day4.txt", "r")
```

```
pairs = input.read().split("\n")
pairs = list(filter(lambda x: len(x) > 0, pairs))
pairs = [[[int(n) for n in side.split("-")] for side in pair.split(",")] for pair in pairs]
```

```
#part1
total = 0
for a,b in pairs:
    if (a[0] <= b[0] and a[1] >= b[1]) or (b[0] <= a[0] and b[1] >= a[1]):
        total+=1
print(total)
```

```
#part2
total = 0
for a,b in pairs:
    if a[0] <= b[1] and b[0] <= a[1]:
        total+=1
print(total)
```
